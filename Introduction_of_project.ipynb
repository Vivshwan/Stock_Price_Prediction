# **Project Summary -**

📌 **Project Overview**

This project develops a deep learning-based solution for predicting Tesla (TSLA) stock prices using historical adjusted closing prices. The system compares two recurrent neural network architectures—SimpleRNN and LSTM (Long Short-Term Memory)—to determine which model better captures temporal patterns in financial time series data. A complete pipeline is implemented, including data preprocessing, model training, evaluation with multiple metrics, and an interactive web dashboard for real-time predictions.

🔧 **Technical Architecture**

**Data Preprocessing Pipeline**

The project begins by loading Tesla stock data from a CSV file containing columns such as Date, Open, High, Low, Close, Adjusted Close, and Volume. The Adjusted Close price is selected as the target feature because it accounts for stock splits and dividends, making it a reliable indicator of true value. Missing values are handled using forward fill followed by backward fill to maintain temporal continuity. The data is then normalized to the [0, 1] range using MinMaxScaler, which helps neural networks converge faster during training.

**Sequence Creation**

The core innovation is transforming the time series into a supervised learning format using a sliding window approach. The model uses 60 consecutive past days (approximately 3 trading months) as input to predict the next day's price. This creates overlapping sequences: days 1-60 predict day 61, days 2-61 predict day 62, and so on. The data is split chronologically (80% training, 20% testing) without random shuffling to prevent data leakage—a critical requirement for time series forecasting.

**Model Architecture**

Two models are implemented and compared:

SimpleRNN: Basic recurrent network with 50 units, prone to vanishing gradient issues for long sequences

LSTM: Advanced architecture with forget, input, and output gates (50 units) designed to retain information over longer periods

Both models include a 20% dropout layer to prevent overfitting and a dense output layer with linear activation for price regression. Training uses the Adam optimizer with mean squared error (MSE) as the loss function. Early stopping (patience of 10 epochs) and model checkpointing ensure only the best model is saved.

📊 **Evaluation Metrics**


Models are evaluated using five metrics on the test set after inverse transformation back to dollar values:

**Metric Interpretation**

MSE	Penalizes large errors heavily
RMSE	Average error in dollars
MAE	Absolute average error in dollars
R² Score	Variance explained (94% for SimpleRNN, 91% for LSTM)
MAPE	Percentage-based error

🖥️ **Interactive Dashboard**

A Streamlit web application provides an intuitive interface for users to:

Select between trained SimpleRNN and LSTM models

Adjust lookback window (10-120 days) and forecast horizon (1-14 days)

View interactive Plotly charts showing historical prices, daily returns, and predictions

Download the dataset and see real-time metrics including current price, volatility, and price changes

🎯 **Key Findings & Limitations**

Finding: SimpleRNN achieved 94% R² score, outperforming LSTM's 91% in this specific configuration.


Limitations: The models use only price data, ignoring trading volume, news sentiment, macroeconomic indicators, and company-specific events. Future improvements could include technical indicators (RSI, MACD), sentiment analysis from social media, and ensemble methods for enhanced accuracy.

# **Problem Statement**


**The Challenge**

Stock price prediction is a complex and challenging task due to the non-linear, volatile, and time-dependent nature of financial markets. Traditional forecasting methods such as linear regression, ARIMA, and moving averages struggle to capture the sequential patterns and long-term dependencies present in stock price movements.

**Specific Problems Addressed**
1. **Temporal Dependency Capture-**
Standard machine learning models treat each day as an independent observation, ignoring the fact that today's price is influenced by past prices, trends, and patterns. This project addresses the need for models with built-in memory that can learn from historical sequences.

2. **Model Selection Dilemma-**
Practitioners often face uncertainty when choosing between different recurrent neural network architectures. There is a need for a comparative analysis of SimpleRNN vs LSTM specifically for financial time series forecasting.

3. **Lack of End-to-End Solutions-**
Many existing solutions focus only on model training, lacking:

        Proper time series preprocessing (chronological splitting, sequence creation)

        Comprehensive evaluation metrics (R², MAPE, RMSE)

        User-friendly interfaces for real-time predictions

4. **Interpretability Gap-**
Technical metrics like MSE are not easily interpretable for business users. Predictions need to be converted back to actual dollar values with clear accuracy measures.

**Project Objectives**

Build a complete pipeline from raw data to prediction visualization

Compare SimpleRNN and LSTM architectures for stock prediction

Provide an interactive dashboard for easy model deployment

Deliver interpretable metrics in both dollar values and percentages


This project bridges the gap between complex deep learning techniques and practical, user-friendly stock price forecasting.
